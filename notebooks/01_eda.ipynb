```python

import pandas as pd
import plotly.express as px
from src.data.loader import load_data 
```

```python
df = load_data()
```

```python
conversion_rate = df["Converted"].value_counts(normalize=True)* 100
```

```python
figure = px.bar(x = ["Not Converted", "Converted"], y = [ conversion_rate[0], conversion_rate[1]], labels = {"x": "Outcome", "y": "Percentage"}, title = "Overall conversion rate")
figure.show()
print("Overall conversion rate: ", conversion_rate[1], "%")

```

```python
columns_to_compare = ["ProductRelated", "BounceRates", "ExitRates", "PageValues"]
```

```python
figure = px.box(
    df.melt(id_vars = "Converted", value_vars = columns_to_compare), 
    x = "variable",
    y = "value",
    color = "Converted",
    title = "Buyers vs non buyers",
    labels = {"variable": "Metric", "value": "Value", "Converted": "Converted" }
)
figure.show()
```

```python
#conversion rate by month
monthly_conversion = df.groupby("Month")["Converted"].mean() * 100

figure = px.bar(x = monthly_conversion.index, y = monthly_conversion.values, labels = {"x": "Month", "y": "Conversion Rate"}, title = "Conversion rate by month")
figure.show()
```

```python
#conversion rate by special day
special_day_conversion = df.groupby("SpecialDay")["Converted"].mean() * 100

figure = px.line(x = special_day_conversion.index, y = special_day_conversion.values, labels = {"x": "Special Day", "y": "Conversion Rate"}, title = "Conversion rate by special day")
figure.show()
```

```python
#Visitor type(returning vs new) and conversion rate
visitor_type_conversion = df.groupby("VisitorType")["Converted"].mean() * 100
figure = px.bar(x = visitor_type_conversion.index, y = visitor_type_conversion.values, labels = {"x": "Visitor Type", "y": "Conversion Rate"}, title = "Conversion rate by visitor type")
figure.show()
```

```python
#Numeric columns moving together
numeric_columns = df.select_dtypes(include=['number'])
corr = numeric_columns.corr()
figure = px.imshow(corr, text_auto= ".2f", title="Correlation Matrix of Numeric Features", color_continuous_scale="RdBu_r", aspect="auto")
figure.show()
```
